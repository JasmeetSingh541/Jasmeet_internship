# Code 1

```python
from scipy import stats
import numpy as np
hours = np.array([1, 2, 3, 4, 5, 6, 7, 8])
marks = np.array([15, 22, 30, 38, 45, 53, 60, 68])
slope, intercept, r, p, std_err = stats.linregress(hours, marks)
print("Slope:", slope)
print("Intercept:", intercept)
print("Correlation coefficient:", r)
study_hours = float(input("Enter hours studied: "))
predicted_marks = slope * study_hours + intercept
print(f"Predicted marks: {predicted_marks:.2f}")
```

# Code 2

```python
from scipy import stats
import numpy as np
area = np.array([500, 650, 800, 950, 1100, 1250, 1400, 1550])
price = np.array([22, 27, 33, 37, 43, 48, 54, 59])
slope, intercept, r, p, std_err = stats.linregress(area, price)
print("Slope:", slope)
print("Intercept:", intercept)
print("Correlation coefficient:", r)
house_area = float(input("Enter house area in square feet: "))
predicted_price = slope * house_area + intercept
print(f"Predicted house price: Rs. {predicted_price:.2f} lakh")
```

# Code 3

```python
import numpy as np
x = np.array([0, 1, 2, 3, 4, 5, 6])
y = np.array([2, 6, 14, 26, 42, 62, 86])
model = np.poly1d(np.polyfit(x, y, 2))
new_x = float(input("Enter an input value: "))
predicted_y = model(new_x)
print(f"Predicted output: {predicted_y:.2f}")
```

# Code 4

```python
from sklearn.neighbors import KNeighborsClassifier

# Training dataset: [Weight, Sweetness]
X = [
    [120, 8.5],
    [130, 8.0],
    [140, 7.8],
    [150, 8.2],
    [160, 7.5],
    [170, 4.0],
    [180, 3.5],
    [190, 4.2],
    [200, 3.8],
    [210, 3.2]
]
y = [
    "Apple", "Apple", "Apple", "Apple", "Apple",
    "Orange", "Orange", "Orange", "Orange", "Orange"
]
# Create and train the KNN model
model = KNeighborsClassifier(n_neighbors=3)
model.fit(X, y)

# Take input from the user
weight = float(input("Enter fruit weight: "))
sweetness = float(input("Enter sweetness level: "))

# Predict the fruit
result = model.predict([[weight, sweetness]])

print("Predicted Fruit:", result[0])
```

# Code 5

```python
from sklearn.neighbors import KNeighborsClassifier

X = [
    [1, 45],
    [2, 50],
    [2, 60],
    [3, 55],
    [3, 65],
    [4, 70],
    [5, 75],
    [6, 80],
    [7, 85],
    [8, 95]
]
y = [
    "Fail", "Fail", "Fail", "Fail", "Fail",
    "Pass", "Pass", "Pass", "Pass", "Pass"
]

model = KNeighborsClassifier(n_neighbors=3)
model.fit(X, y)

hours = float(input("Enter study hours per day: "))
attendance = float(input("Enter attendance percentage: "))

result = model.predict([[hours, attendance]])

print("Predicted Result:", result[0])
```
